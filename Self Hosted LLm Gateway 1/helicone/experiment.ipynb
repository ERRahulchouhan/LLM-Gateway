# Helicone deployment smoke test

After deploying Helicone with its official self-hosted stack, set `HELICONE_BASE_URL` to the configured instance. This checks the root endpoint only; use the upstream docs for authenticated gateway/observability tests.

```python
import os
from urllib.request import urlopen

base_url = os.getenv("HELICONE_BASE_URL")
if not base_url:
    raise RuntimeError("Set HELICONE_BASE_URL to your self-hosted Helicone URL.")
with urlopen(base_url, timeout=15) as response:
    print("HTTP status:", response.status)
    print(response.read(500).decode("utf-8", errors="replace"))
```
