# Portkey Gateway reachability test

Optional smoke test. Start this folder's Compose stack first. Set `GATEWAY_BASE_URL` if using a different local port. This checks reachability only; configure provider and access credentials separately.

```python
import os
from urllib.request import urlopen

base_url = os.getenv("GATEWAY_BASE_URL", "http://127.0.0.1:8788").rstrip("/")
with urlopen(base_url, timeout=10) as response:
    print("HTTP status:", response.status)
    print(response.read(500).decode("utf-8", errors="replace"))
```
