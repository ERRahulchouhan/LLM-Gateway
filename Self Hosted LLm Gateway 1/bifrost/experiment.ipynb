# Bifrost API smoke test

Optional local experiment. Start the Compose stack first. Set `GATEWAY_API_KEY` if gateway authentication is enabled; do not save real keys in this notebook.

```python
import json
import os
from urllib.request import Request, urlopen

base_url = os.getenv("GATEWAY_BASE_URL", "http://127.0.0.1:8080").rstrip("/")
api_key = os.getenv("GATEWAY_API_KEY")
headers = {"Authorization": f"Bearer {api_key}"} if api_key else {}
request = Request(f"{base_url}/v1/models", headers=headers)
with urlopen(request, timeout=15) as response:
    print("HTTP status:", response.status)
    models = json.loads(response.read())
print(json.dumps(models, indent=2))
```
