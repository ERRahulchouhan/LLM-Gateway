# Envoy AI Gateway client smoke test

Set `GATEWAY_BASE_URL` to a reachable deployed Envoy AI Gateway route. Set `GATEWAY_API_KEY` if the route requires a bearer token. This notebook does not deploy the Kubernetes control plane.

```python
import os
from urllib.request import Request, urlopen

base_url = os.getenv("GATEWAY_BASE_URL")
if not base_url:
    raise RuntimeError("Set GATEWAY_BASE_URL to your deployed gateway endpoint.")
headers = {}
api_key = os.getenv("GATEWAY_API_KEY")
if api_key:
    headers["Authorization"] = f"Bearer {api_key}"
with urlopen(Request(base_url, headers=headers), timeout=15) as response:
    print("HTTP status:", response.status)
    print(response.read(500).decode("utf-8", errors="replace"))
```
