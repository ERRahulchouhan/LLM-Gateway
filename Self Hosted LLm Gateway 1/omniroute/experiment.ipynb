# OmniRoute reachability test

Optional smoke test. Start the stack and finish its setup first. Set `GATEWAY_BASE_URL` to the configured service URL. This tests reachability only.

```python
import os
from urllib.request import urlopen

base_url = os.getenv("GATEWAY_BASE_URL", "http://127.0.0.1:20128").rstrip("/")
with urlopen(base_url, timeout=15) as response:
    print("HTTP status:", response.status)
    print(response.read(500).decode("utf-8", errors="replace"))
```
