# New API smoke test

Optional local experiment. Start the Compose stack and create an API token in New API first. Set `GATEWAY_API_KEY` in the notebook environment; do not save real keys in this notebook.

```python
import json
import os
from urllib.request import Request, urlopen

base_url = os.getenv("GATEWAY_BASE_URL", "http://127.0.0.1:3001").rstrip("/")
api_key = os.getenv("GATEWAY_API_KEY")
if not api_key:
    raise RuntimeError("Set GATEWAY_API_KEY in the notebook environment before running this test.")

request = Request(f"{base_url}/v1/models", headers={"Authorization": f"Bearer {api_key}"})
with urlopen(request, timeout=15) as response:
    print("HTTP status:", response.status)
    models = json.loads(response.read())
print(json.dumps(models, indent=2))
```
