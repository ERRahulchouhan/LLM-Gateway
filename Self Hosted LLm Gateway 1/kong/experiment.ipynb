# Kong Gateway status check

Optional local smoke test. Start this folder's Compose stack first. This queries the local Admin API, which must remain private and must not be exposed publicly.

```python
import json
import os
from urllib.request import urlopen

base_url = os.getenv("KONG_ADMIN_URL", "http://127.0.0.1:8001").rstrip("/")
with urlopen(f"{base_url}/status", timeout=10) as response:
    print("HTTP status:", response.status)
    print(json.dumps(json.loads(response.read()), indent=2))
```
