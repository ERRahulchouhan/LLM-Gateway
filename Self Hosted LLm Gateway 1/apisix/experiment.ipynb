# APISIX OpenAI-compatible models test

Optional smoke test. Start this folder's Compose stack and set `APISIX_GATEWAY_KEY` in the notebook environment. This lists upstream models without sending a completion.

```python
import json
import os
from urllib.request import Request, urlopen

base_url = os.getenv("GATEWAY_BASE_URL", "http://127.0.0.1:9082").rstrip("/")
gateway_key = os.getenv("APISIX_GATEWAY_KEY")
if not gateway_key:
    raise RuntimeError("Set APISIX_GATEWAY_KEY in the notebook environment.")
request = Request(f"{base_url}/v1/models", headers={"apikey": gateway_key})
with urlopen(request, timeout=15) as response:
    print("HTTP status:", response.status)
    print(json.dumps(json.loads(response.read()), indent=2))
```
